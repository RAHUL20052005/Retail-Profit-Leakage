```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

seg_orders = Sales_Transactions[Sales_Transactions["Order_Status"] == "Delivered"].copy()
seg_orders = seg_orders.dropna(subset=["Customer_ID", "Selling_Price", "Discount_Amount"])
seg_orders["Gross"] = seg_orders["Quantity"] * seg_orders["Selling_Price"]
seg_orders["Discount_Amount"] = seg_orders["Discount_Amount"].clip(lower=0, upper=seg_orders["Gross"])
seg_orders["Net_Sales"] = (seg_orders["Gross"] - seg_orders["Discount_Amount"]).clip(lower=0)

returned = Returns[Returns["Return_Status"] == "Approved"].groupby("Order_ID").size().clip(upper=1).rename("Returned")
seg_orders = seg_orders.merge(returned, on="Order_ID", how="left").fillna({"Returned": 0})

seg_cust = seg_orders.groupby("Customer_ID").agg(
    Total_Spend=("Net_Sales", "sum"), Orders=("Order_ID", "count"), Returns=("Returned", "sum"),
    Discount=("Discount_Amount", "sum"), Gross=("Gross", "sum")).reset_index()
seg_cust["AOV"] = seg_cust["Total_Spend"] / seg_cust["Orders"]
seg_cust["Return_Freq"] = seg_cust["Returns"] / seg_cust["Orders"]
seg_cust["Discount_Share"] = 100 * seg_cust["Discount"] / seg_cust["Gross"].replace(0, np.nan)

cols = ["Total_Spend", "AOV", "Orders", "Return_Freq", "Discount_Share"]
repeat = seg_cust[seg_cust["Orders"] > 1].copy()
repeat[cols] = repeat[cols].replace([np.inf, -np.inf], np.nan).fillna(0)      # no NaN can survive

X = repeat[cols].clip(lower=0).copy()                                          # no negatives, so log1p is safe
X["Total_Spend"], X["AOV"] = np.log1p(X["Total_Spend"]), np.log1p(X["AOV"])
Z = StandardScaler().fit_transform(X)
assert not np.isnan(Z).any(), "still has NaN"                                  # clear message if something remains

repeat["Cluster"] = KMeans(n_clusters=4, n_init=20, random_state=7).fit_predict(Z)
profile = repeat.groupby("Cluster")[cols].mean().round(2)
profile["Customers"] = repeat.groupby("Cluster").size()
print(profile)
```

```python
names = {}
names[profile["Return_Freq"].idxmax()] = "High return risk"
rest = profile.drop(index=list(names))
names[rest["Discount_Share"].idxmax()] = "Deal hunters"
rest = profile.drop(index=list(names))
names[rest["Total_Spend"].idxmax()] = "High value loyal"
for c in profile.index:
    names.setdefault(c, "Regular full-price buyers")

repeat["Segment"] = repeat["Cluster"].map(names)
seg_cust["Segment"] = seg_cust["Customer_ID"].map(repeat.set_index("Customer_ID")["Segment"]).fillna("One-time buyers")
print(seg_cust["Segment"].value_counts())
seg_cust.to_excel("Customer_Segments.xlsx", index=False)
```

```python
import numpy as np, pandas as pd

p = Product_Master[["SKU_ID", "SKU_Name", "Category", "Unit_Cost", "MRP"]]
sales_d = Sales_Transactions[Sales_Transactions["Order_Status"] == "Delivered"].dropna(subset=["Selling_Price", "Discount_Amount"]).merge(p, on="SKU_ID")
sales_d["Gross"] = sales_d["Quantity"] * sales_d["Selling_Price"]
sales_d["Net"] = sales_d["Gross"] - sales_d["Discount_Amount"]
sales_d["COGS"] = sales_d["Quantity"] * sales_d["Unit_Cost"]
sales_d["Profit"] = sales_d["Net"] - sales_d["COGS"]

ref_ok = Returns[Returns["Return_Status"] == "Approved"]

# ---- 1. Top 5 loss drivers (Rs) ----
cancelled = Sales_Transactions[Sales_Transactions["Order_Status"] == "Cancelled"].dropna(subset=["Selling_Price", "Discount_Amount"])
cancelled_lost = (cancelled["Quantity"] * cancelled["Selling_Price"] - cancelled["Discount_Amount"]).sum()
drivers = pd.DataFrame({
    "Loss_Driver": ["Discounts given", "Approved returns / refunds", "Sold below cost (margin destroyed)", "Cancelled orders (lost sales)", "Refunds with no matching sale"],
    "Amount": [sales_d["Discount_Amount"].sum(), ref_ok["Refund_Amount"].sum(), -sales_d.loc[sales_d["Profit"] < 0, "Profit"].sum(),
               cancelled_lost, ref_ok.loc[~ref_ok["Order_ID"].isin(Sales_Transactions["Order_ID"]), "Refund_Amount"].sum()]
}).sort_values("Amount", ascending=False).head(5).round(0)

# ---- 2. Worst 3 vendors ----
v = Vendor_Supply.copy()
v["Delay"] = (pd.to_datetime(v["Actual_Delivery_Date"]) - pd.to_datetime(v["Expected_Delivery_Date"])).dt.days
v = v[v["Delay"].abs() < 20]
vendors = v.groupby("Vendor_ID").agg(POs=("PO_ID", "count"), Avg_Delay_Days=("Delay", "mean"),
                                     SLA_Breach_Pct=("Delay", lambda x: 100 * (x > 0).mean())).round(2)
worst_vendors = vendors[vendors["POs"] >= 100].sort_values("SLA_Breach_Pct", ascending=False).head(3).reset_index()

# ---- 3. Stockout hotspots (SKU-store with most zero-stock days) ----
hot = (Inventory_Daily[Inventory_Daily["Closing_Stock"] == 0].groupby(["SKU_ID", "Store_ID"]).size()
       .rename("Stockout_Days").reset_index().sort_values("Stockout_Days", ascending=False).head(10))

# ---- 4. Best 5 SKUs to promote: high margin, low returns ----
sku = sales_d.groupby("SKU_ID").agg(Units=("Quantity", "sum"), Net=("Net", "sum"), Profit=("Profit", "sum"), Lines=("Order_ID", "count")).reset_index()
rets = ref_ok.groupby("SKU_ID").size().rename("Returns")
sku = sku.merge(rets, on="SKU_ID", how="left").fillna({"Returns": 0})
sku["Margin_Pct"] = 100 * sku["Profit"] / sku["Net"]
sku["Return_Rate_Pct"] = 100 * sku["Returns"] / sku["Lines"]
sku = sku[sku["Lines"] >= sku["Lines"].median()]                       # enough sales to trust
sku["Score"] = sku["Margin_Pct"] - 2 * sku["Return_Rate_Pct"]
best = sku.sort_values("Score", ascending=False).head(5).merge(p[["SKU_ID", "SKU_Name", "Category"]], on="SKU_ID").round(2)

# ---- export to Excel ----
with pd.ExcelWriter("Executive_Summary_Report.xlsx") as w:
    drivers.to_excel(w, sheet_name="Top5_Loss_Drivers", index=False)
    worst_vendors.to_excel(w, sheet_name="Worst3_Vendors", index=False)
    hot.to_excel(w, sheet_name="Stockout_Hotspots", index=False)
    best.to_excel(w, sheet_name="Best5_SKUs_To_Promote", index=False)

# ---- export to PDF ----
from reportlab.lib.pagesizes import A4
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.platypus import SimpleDocTemplate, Paragraph, Table, TableStyle, Spacer

def tbl(df):
    t = Table([list(df.columns)] + df.astype(str).values.tolist())
    t.setStyle(TableStyle([("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#1F3A5F")), ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
                           ("FONTSIZE", (0, 0), (-1, -1), 8), ("GRID", (0, 0), (-1, -1), 0.25, colors.grey)]))
    return t

st = getSampleStyleSheet()
doc = SimpleDocTemplate("Executive_Summary_Report.pdf", pagesize=A4)
parts = [Paragraph("Retail Chain Profit Leakage - Executive Summary", st["Title"])]
for title, df in [("Top 5 loss drivers (Rs)", drivers), ("Worst 3 vendors", worst_vendors),
                  ("Stockout hotspots", hot.head(5)), ("Best 5 SKUs to promote", best[["SKU_ID", "SKU_Name", "Category", "Margin_Pct", "Return_Rate_Pct"]])]:
    parts += [Paragraph(title, st["Heading3"]), tbl(df), Spacer(1, 12)]
doc.build(parts)
print(drivers, worst_vendors, hot.head(5), best, sep="\n\n")
```
